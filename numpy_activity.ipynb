```python
#Activity 4

Temperature = np.array([
    27.0, 27.5, 28.0, 27.2, 26.8, 27.4, 27.9, 28.1, 27.5, 27.0,
    26.9, 27.3, 27.8, 28.2, 27.7, 27.1, 26.8, 27.2, 27.6, 28.0,
    27.4, 27.0, 26.7, 27.1, 27.5, 27.9, 28.3, 27.8, 27.2, 26.9, 27.3
])

indices = np.where(Temperature == np.max(Temperature))[0]
print(" the hottest day of the month is",indices+1,"July")
print(" ")

indices2 = np.where(Temperature == np.min(Temperature))[0]
print("The Coldest day of month is ", indices2 +1,"July")
print(" ")

print("The Average Temperature of month is ", np.mean(Temperature))
print(" ")


indices3 = np.where(Temperature>30)[0]
print(indices3)
```
