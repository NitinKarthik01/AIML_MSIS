```python
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)

df = iris.frame

df.head()
```

```python
# Activity 2

plt.scatter(df['sepal length (cm)'], df['petal length (cm)'])
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.show()
```

```python
#Activity 3

plt.hist(df['petal width (cm)'])

plt.title("Distribution")
plt.show()
```

```python
#Activity 4

species = ['Setosa','Versicolor','Verginica']

count = df['target'].value_counts()

plt.bar(species,count)

plt.show()
```

```python
# Assignment

days = list(range(1, 32))

Hyd = [
    31, 32, 32, 29, 29, 33, 33, 33, 34, 35,
    34, 35, 37, 36, 35, 35, 31, 32, 33, 32,
    29, 28, 31, 31, 31, 32, 31, 32, 26, 26, 29
]
udupi = [
    26, 28, 28, 29, 30, 30, 30, 30, 31, 30,
    30, 30, 30, 31, 31, 30, 27, 29, 29, 27,
    29, 26, 29, 30, 29, 29, 30, 28, 29, 30, 30
]
plt.plot(days,Hyd,udupi)
plt.title("Temp of Hyd vs Udupi in July")
plt.xlabel("Days of July")
plt.ylabel("Temperature in C")
plt.legend(['Hyd','Udupi'])
plt.show()
```

```python

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]


hyd_high_temperatures= [
    29, 32, 36, 38, 39, 33,
    29, 29, 29, 30, 29, 28
]

hyd_low_temperatures = [
    18, 20, 23, 27, 28, 26,
    23, 23, 22, 21, 19, 18
]


udupi_high_temperatures = [
    32, 32, 33, 35, 34, 29,
    27, 28, 29, 30, 32, 31
]

udupi_low_temperatures = [
    22, 22, 23, 24, 25, 24,
    23, 23, 23, 23, 23, 22
]

plt.figure(figsize=(7.15, 5))
plt.plot(months,hyd_high_temperatures, label = "Hyd_high")
plt.plot(months,hyd_low_temperatures, label = "Hyd_Low")

plt.plot(months,udupi_high_temperatures, label = "Udupi_high")
plt.plot(months,udupi_low_temperatures, label = "Udupi_low")
plt.title("Temp of Hyd vs Udupi in 2025")
plt.xlabel("Months")
plt.ylabel("Temperature in C")
plt.legend()

plt.show()
```
