```python
#Assignment

# total no.of passengers
# find avg age of survivers and non survivers
#list the passengers who are older thann 60 years
#find the avg fare of each passenger class

file_path = r"/content/drive/MyDrive/titanic.csv"
df = pd.read_csv(file_path)
df.head()

print("The no.of Passengers on Titanic are",len(df))
print(" ")
print("The average age of Survivers",df.loc[df['Survived'] == 1, 'Age'].mean() )
print(" ")
print("The average age of Non - Survivers",df.loc[df['Survived'] == 0, 'Age'].mean() )
print(" ")
print("The average fare of CLASS 1 ",df.loc[df['Pclass'] == 1, 'Fare'].mean() )
print(" ")
print("The average fare of CLASS 2 ",df.loc[df['Pclass'] == 2, 'Fare'].mean() )
print(" ")
print("The average fare of CLASS 3 ",df.loc[df['Pclass'] == 3, 'Fare'].mean() )
print(" ")
x = df[df['Age'] > 60]
print("The lis of passengers above 60 \n",x)

```
