```python
# lambda is a one line function

def ho_add(x):
  return lambda y: x+y

# "x" in lambda function is bounded to "x" isn parent function, lamda can see all variables of parent function, it is called visibility.
```

```python
ho_add
```

```python
f = ho_add(100)
```
